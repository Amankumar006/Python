```python
import time
import os

grid_size = 5

start = (0, 0)
goal = (4, 4)

obstacles = [
    (0, 2), (1, 2),
    (2, 2), (3, 1), (3, 3)
]

path = [
    (0, 0), (1, 0),(2, 0),
    (2, 1), (2, 2),(2, 3), 
    (3, 3),(4, 3),(4, 4)
]

def create_grid(agent_position):
    grid = [
        ['.' for _ in range(grid_size)]
        for _ in range(grid_size)
    ]

    for obstacle in obstacles:
        grid[obstacle[0]][obstacle[1]] = '#'

    grid[goal[0]][goal[1]] = 'G'
    grid[agent_position[0]][agent_position[1]] = 'A'

    return grid

def display_grid(grid):
    for row in grid:
        print(' '.join(row))
    print()

for position in path:
    os.system('cls' if os.name == 'nt' else 'clear')

    grid = create_grid(position)
    display_grid(grid)

    if position == goal:
        print("Goal reached!")
        break

    time.sleep(0.5)
```
