# Program 2: Rule-Based Learning (CN2 & FOIL)
**Aim:** Read a dataset and use an example-based method (such as RIPPER or CN2) to generate a set of classification rules. Apply the FOIL algorithm (First-Order Inductive Learner) to learn first-order rules for predicting.

```python
# Install required package: pip install Orange3
from Orange.classification import CN2Learner
from Orange.data import Table

# Load a sample dataset (Iris)
data = Table("iris")

# Apply CN2 Algorithm
learner = CN2Learner()
classifier = learner(data)

print("CN2 Classification Rules:")
for rule in classifier.rule_list:
    print(rule)
```

```python
# Simple FOIL-style example
def foil_algorithm(examples, target_attr):
    rules = []
    positives = [ex for ex in examples if ex[target_attr] == "Yes"]

    while positives:
        rule = []
        covered = examples[:]
        while any(ex[target_attr] == "No" for ex in covered):
            # Pick literal from positives that excludes negative examples
            for attr in [k for k in examples[0] if k != target_attr]:
                val = positives[0][attr]
                new_covered = [ex for ex in covered if ex[attr] == val]
                if not any(ex[target_attr] == "No" for ex in new_covered):
                    rule.append((attr, val))
                    covered = new_covered
                    break
            break
        rules.append(rule)
        positives = [ex for ex in positives if ex not in covered]
    return rules

# Example dataset from lab manual
examples = [
    {"Outlook": "Sunny", "Wind": "Strong", "Play": "No"},
    {"Outlook": "Sunny", "Wind": "Weak", "Play": "Yes"},
    {"Outlook": "Overcast", "Wind": "Weak", "Play": "Yes"},
    {"Outlook": "Rainy", "Wind": "Weak", "Play": "Yes"},
]

foil_rules = foil_algorithm(examples, "Play")
print("FOIL Rules:")
for r in foil_rules:
    conds = " AND ".join([f"{a}={v}" for a, v in r])
    print(f"IF {conds} THEN Play=Yes")
```
